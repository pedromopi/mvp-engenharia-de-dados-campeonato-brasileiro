## Aplicação dos comentários

```python
def texto_sql(texto):
    return texto.replace("'", "''")

for tabela, dados in catalogo.items():
    spark.sql(f"COMMENT ON TABLE {tabela} IS '{texto_sql(dados['descricao'])}'")
    for coluna, descricao in dados['colunas'].items():
        spark.sql(
            f"ALTER TABLE {tabela} ALTER COLUMN `{coluna}` COMMENT '{texto_sql(descricao)}'"
        )

print(f"{len(catalogo)} tabelas documentadas.")
```

## Conferência

```python
linhas = []

for tabela in catalogo:
    for coluna in spark.catalog.listColumns(tabela):
        linhas.append((
            tabela, coluna.name, coluna.dataType, coluna.description or ''
        ))

resultado = spark.createDataFrame(
    linhas, ['tabela', 'coluna', 'tipo', 'descricao']
)
display(resultado.orderBy('tabela', 'coluna'))
```
